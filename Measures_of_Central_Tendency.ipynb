```python
import seaborn as sns
import pandas as pd
import numpy as np

# Load the tips dataset
tips = sns.load_dataset('tips')
```

```python
tips.head(100)
```

```python
tips.info()
```

```python
total_bill_mean = np.mean(tips['total_bill'])

print(f"mean of bills: {total_bill_mean}")
```

```python
total_tip_mean = tips['tip'].mean()

print(f"mean_of_tips: {total_tip_mean}")
```

```python
total_bill_median = np.median(tips["total_bill"])

print(f"median_of_bills: {total_bill_median}")
```

```python
tip_median = tips["tip"].median()

print(f"median_of_tips: {tip_median}")
```

```python
tips['tip'].value_counts()
```

```python
tip_mode = tips['tip'].mode()

print(f"mode of tips: {tip_mode[0]}")
```

```python
mode_sex = tips['sex'].mode()

print(f"mode_of_sex: {mode_sex[0]}")
```

```python
mode_day = tips['day'].mode()

print(f"mode_of_day: {mode_day[0]}")
```

```python
mode_time = tips['time'].mode()

print(f"mode_of_time: {mode_time[0]}")
```

**Starting with Measures of Dispersion:**

```python
total_bill_range = tips['total_bill'].max() - tips['total_bill'].min()

print(f"Range of total_bill = {total_bill_range}")
print(tips['total_bill'].max())
print(tips['total_bill'].min())
```

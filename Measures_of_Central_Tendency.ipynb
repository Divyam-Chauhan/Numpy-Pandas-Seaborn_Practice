# **Starting with Measures of Dispersion:**

**Range:**

```python
total_bill_range = tips['total_bill'].max() - tips['total_bill'].min()

print(f"Range of total_bill = {total_bill_range}")
print(f"Maximum bill = {tips['total_bill'].max()}")
print(f"Minimum bill = {tips['total_bill'].min()}")
```

```python
total_tips_range = tips['tip'].max() - tips['tip'].min()

print(f"Range of total tips = {total_tips_range}")
print(f"Maximum tip = {tips['tip'].max()}")
print(f"Minimum tip = {tips['tip'].min()}")
```

# Variance:

Pandas Variance and Numpy variance is slightly different because pandas finds Sample variance (denominator is n-1) and numpy finds Variance (denominator is n)

Variance = σ² = Σ(Xᵢ − μ)²

Where:
- σ² = Variance
- n = Total number of observations
- μ = Mean

Sample Variance = s² = Σ(Xᵢ − x̅)² / (n-1)

Where:
- s² = Variance
- n = Total number of observations
- x̅ = Mean of the sample

```python
np_total_bill_variance = np.var(tips['total_bill'])
print(f"Numpy Variance of total_bill = {np_total_bill_variance}")
```

```python
pd_total_bill_variance = tips['total_bill'].var()
print(f"pandas Variance of total bill = {pd_total_bill_variance}")
```

# Standard Deviation

```python
tips_std = tips['tip'].std()
print(f"Standard Deviation of tips = {tips_std}")
```

```python
np_tips_std = np.std(tips['tip'])
print(f"Numpy Standard Deviation of tips = {np_tips_std}")
```

# Interquartile Range (IQR):

```python
q1_totalBill = tips['total_bill'].quantile(0.25)
q2_totalBill = tips['total_bill'].quantile(0.50)
q3_totalBill = tips['total_bill'].quantile(0.75)

print(f"First Quartile of total Bill = {q1_totalBill}")
print(f"Second Quartile/Median of Total Bill = {q2_totalBill}")
print(f"Third Quartile of Total Bill = {q3_totalBill}")
```

```python
q1_tips = tips['tip'].quantile(0.25)
q2_tips = tips['tip'].quantile(0.50)
q3_tips = tips['tip'].quantile(0.75)

print(f"First Quartile of tips = {q1_tips}")
print(f"Second Quartile/Median of tips = {q2_tips}")
print(f"Third Quartile of Tips = {q3_tips}")
```
